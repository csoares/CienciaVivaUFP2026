Para que um carro autónomo consiga conduzir sozinho, ele precisa de **ver** o mundo à sua volta. O nosso JetRacer está equipado com uma câmara grande angular na parte frontal.

Neste exercício, vamos criar um fluxo de vídeo em tempo real diretamente no vosso Jupyter Notebook. 

---

### 🎯 O Teu Objetivo
Ligar a câmara do robô, transmitir as imagens para o ecrã do computador e aprender a desligá-la corretamente para libertar a memória do sistema.

### 🛠️ Instruções Passo a Passo

1. **Aponte o robô:** Coloque o JetRacer virado para a pista ou para um objeto colorido na mesa.
2. **Executa a célula:** Seleciona a célula de código abaixo e prime `Shift + Enter`.
3. **Observa o ecrã:** Vai aparecer uma janela quadrada preta que, passados alguns segundos, começará a mostrar o que o robô está a ver.
4. **Termina a sessão:** Quando terminares de ver, clica no botão vermelho **❌ DESLIGAR CÂMARA**. Isto é muito importante para que a câmara fique livre para o próximo exercício!

---

### 🚀 Desafio de Exploração
No código abaixo, encontras as variáveis `width=300` e `height=300` dentro de `cv2.resize`.
* Estas variáveis controlam a **resolução** (o tamanho) da imagem que é processada.
* O que achas que acontece se mudares ambas para `150`? E para `500`?
* *Nota do Professor:* Imagens maiores dão mais detalhe, mas fazem a IA processar mais devagar!

```python
import ipywidgets as widgets
from IPython.display import display
import traitlets # <--- IMPORTANTE: Usar a biblioteca de links da NVIDIA
from jetcam.csi_camera import CSICamera
from jetcam.utils import bgr8_to_jpeg # <--- Converte a imagem para o formato correto do browser

print("--- INSTANCIAÇÃO DA CÂMARA (NATIVO CORRIGIDO) ---")

# 1. Inicializar a câmara
camera = CSICamera(width=300, height=300, capture_width=1280, capture_height=720, capture_fps=21, flip_method=0)

# 2. Criar o widget da imagem
imagem_widget = widgets.Image(format='jpeg', width=300, height=300)

# 3. A SOLUÇÃO: Criar um link dinâmico que converte a imagem automaticamente para JPEG a caminho do ecrã
camera_link = traitlets.dlink((camera, 'value'), (imagem_widget, 'value'), transform=bgr8_to_jpeg)

# 4. Criar o botão de desligar
botao_desligar = widgets.Button(description="❌ DESLIGAR FEED", button_style='danger')

display(imagem_widget)
display(botao_desligar)

# 5. Função para desligar de forma limpa
def desligar_camera(b):
    camera_link.unlink() # Destrói o link visual
    camera.running = False # Desliga o sensor físico
    botao_desligar.description = "🎥 FEED DESLIGADO"
    botao_desligar.button_style = "info"
    botao_desligar.disabled = True
    print("\n🛑 Link destruído e câmara libertada com sucesso!")

botao_desligar.on_click(desligar_camera)

# 6. Ativar o fluxo em segundo plano da câmara
camera.running = True
print("Câmara a transmitir de forma segura e assíncrona!")
```
