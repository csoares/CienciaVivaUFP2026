# 📅 Dia 2 - Exercício 4: O "Cruise Control" Automático

Neste exercício, vamos aprender a controlar o Jetracer diretamente de dois botões no notebook.

1. Executa a célula abaixo.
2. Clica no botão verde **🟢 LIGAR CRUISE** para iniciar o movimento.
3. Clica no botão vermelho **🛑 PARAR CARRO** para interromper o movimento.
4. Experimenta alterar o valor da variável `velocidade` para valores entre `0.1` e `0.25` e testa novamente.

```python
import ipywidgets as widgets
from IPython.display import display
from jethelper import mover, parar

print("--- PAINEL DE CONTROLO DO JETRACER ---")

# 1. Criar os botões
botao_ligar = widgets.Button(description="🟢 LIGAR CRUISE", button_style='success')
botao_parar = widgets.Button(description="🛑 PARAR CARRO", button_style='danger')

# Colocar os botões lado a lado na tela
interface = widgets.HBox([botao_ligar, botao_parar])
display(interface)

# 2. Funções de ação (Eventos)
def ligar_cruise(b):
    print("Cruise Control Ativado: Velocidade 0.15")
    mover(0.15, 0.0) # O carro liga o motor e mantém a velocidade sozinho

def desligar_cruise(b):
    print("Cruise Control Desativado: Parando...")
    parar()

# 3. Ligar os botões às funções
botao_ligar.on_click(ligar_cruise)
botao_parar.on_click(desligar_cruise)
```
