# 📅 Dia 2 - Exercício 4: O Acelerador Virtual 🕹️

Nos videojogos ou quando conduzimos um carro real, a velocidade não é apenas "ligada" ou "desligada". Nós controlamos a aceleração de forma **gradual** (com mais ou menos força).

Neste exercício, vamos criar um **Slider (Barra Deslizante)** que vai funcionar exatamente como o pedal do acelerador do JetRacer!

---

### 🎯 O Teu Objetivo
Controlar a velocidade e o sentido do robô (frente e marcha-atrás) em tempo real, arrastando uma barra com o rato.

### 🛠️ Instruções Passo a Passo

1. **Segurança Primeiro:** Confirma que o teu JetRacer continua com as **rodas no ar** (em cima do bloco).
2. **Executa a célula:** Seleciona a célula de código abaixo e prime `Shift + Enter`.
3. **Usa o Acelerador:** 
   * Arrasta o slider devagar para a **direita** (valores positivos) ➔ As rodas começam a girar para a **frente**.
   * Arrasta o slider para a **esquerda** (valores negativos) ➔ As rodas giram em **marcha-atrás**.
   * Larga o slider no **0.0** ➔ O carro **pára** imediatamente.

---

### 🚀 Desafio de Exploração
Olha para o código abaixo e descobre onde estão definidos os limites do slider:
* `min=-0.3` (Marcha-atrás máxima)
* `max=0.3` (Velocidade máxima para a frente)

Se quisesses dar mais potência ao carro para ele conseguir subir uma rampa na pista real, para que valores mudarias o `max`? Experimenta alterar no código (por exemplo, para `0.4`), executa a célula de novo e sente a diferença na força dos motores!

```python
import ipywidgets as widgets
from IPython.display import display
from jethelper import mover, parar

print("--- PEDAL DO ACELERADOR VIRTUAL ATIVO ---")

# 1. Criação do componente visual (Slider)
acelerador = widgets.FloatSlider(
    value=0.0,              # O carro começa parado (0.0)
    min=-0.3,               # Limite máximo de marcha-atrás (-30%)
    max=0.3,                # Limite máximo de frente (+30%)
    step=0.05,              # O slider salta de 0.05 em 0.05
    description='Acelerar:',
    continuous_update=True  # IMPORTANTE: Envia o sinal enquanto estás a arrastar!
)

display(acelerador)

# 2. Função que escuta o slider e atualiza os motores da Jetson Nano
def atualizar_motor(change):
    nova_velocidade = change['new']
    print(f"Velocidade atual do motor: {nova_velocidade:.2f} ", end='\r')
    mover(nova_velocidade, 0.0)

# 3. Dizemos ao Jupyter para monitorizar o valor do slider e chamar a função acima
acelerador.observe(atualizar_motor, names='value')
```
