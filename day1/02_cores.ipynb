As cores que uma camara ve sao sempre uma mistura de vermelho, verde e azul
(por isso se chama "RGB" - Red, Green, Blue).

Por exemplo:
- **Amarelo** = muito vermelho + muito verde (pouco azul)
- **Roxo / Magenta** = muito vermelho + muito azul (pouco verde)
- **Ciano** = muito verde + muito azul (pouco vermelho)

O formador vai dar-te um objeto **amarelo**. Tira uma foto dele e olha para
os 3 valores. Confirma se a tua previsao acima estava certa.

```python
foto = robot.tirar_foto()
robot.mostrar_foto(foto)

print("Vermelho:", robot.quanto_vermelho(foto))
print("Verde:", robot.quanto_verde(foto))
print("Azul:", robot.quanto_azul(foto))
print("Cor dominante (segundo a funcao):", robot.cor_dominante(foto))
```
