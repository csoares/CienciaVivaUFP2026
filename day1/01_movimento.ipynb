## As funcoes que tens disponiveis

| Funcao | O que faz | Exemplo |
|---|---|---|
| `robot.parar()` | Para o robot imediatamente | `robot.parar()` |
| `robot.andar_frente(segundos)` | Anda em frente, em linha reta | `robot.andar_frente(1)` |
| `robot.andar_tras(segundos)` | Anda para tras, em linha reta | `robot.andar_tras(1)` |
| `robot.virar_esquerda(segundos)` | Anda a virar para a esquerda | `robot.virar_esquerda(0.5)` |
| `robot.virar_direita(segundos)` | Anda a virar para a direita | `robot.virar_direita(0.5)` |

O numero entre parentesis e' sempre o **tempo em segundos** que o robot vai
fazer aquele movimento. Por exemplo, `robot.andar_frente(2)` anda em frente
durante 2 segundos e depois para sozinho.
